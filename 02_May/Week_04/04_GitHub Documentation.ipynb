# Part 1: README Template

Every project README follows this structure:

```markdown
# Project Name

One-sentence description of what this project does.

![Demo](demo.gif)

## What It Does
- Bullet 1
- Bullet 2
- Bullet 3

## How It Works
1. Step 1 (brief)
2. Step 2
3. Step 3

## How to Run
```bash
pip install -r requirements.txt
python main.py --input path/to/input
```

## Results
| Metric | Value |
|--------|-------|
| Accuracy | 94% |
| FPS | 28 |

## What I Learned
2-3 sentences summarising the key insight.
```

Keep it short. A recruiter spends 15 seconds on a repo. Make it immediately clear what it does.

````python
# ── Generate README files for all Phase 1+2 projects ─────────────────────
# Run this cell to create template README.md files in each project folder.
# Then fill in the project-specific details.

import os

PROJECTS = [
    {
        'path':  r'c:/Users/User/Desktop/CV_Practice/02_May/Week_04',
        'name':  'Vehicle Counter + Speed Estimator',
        'desc':  'Real-time vehicle counter and speed estimator using background subtraction, Kalman tracking, and perspective calibration.',
        'does':  ['Counts vehicles crossing a virtual line from a fixed camera',
                  'Estimates speed of each vehicle in km/h',
                  'Reports traffic flow in vehicles per minute',
                  'Saves speed log to CSV'],
        'how':   ['Background subtraction (MOG2) detects moving vehicles',
                  'Contour detection finds the centroid of each vehicle',
                  'Kalman filter assigns stable IDs and smooths positions',
                  'Perspective transform converts pixels to metres for speed calculation',
                  'Counting line detects each vehicle exactly once'],
        'run':   'python main.py --source road.mp4 --line-y 300 --mpp 0.05',
        'metrics': [('FPS (720p)', '~28'), ('Count accuracy', '>95%'), ('Speed error', '<5 km/h')],
        'learned': 'I learned that pixel distance alone is meaningless for speed — you need to calibrate the scene using known real-world measurements. The Kalman filter velocity state gives a much smoother speed estimate than frame-to-frame distance.'
    },
]

README_TEMPLATE = '''# {name}

{desc}

## What It Does
{does}

## How It Works
{how}

## How to Run
```bash
pip install -r requirements.txt
{run}
```

## Results
{metrics}

## What I Learned
{learned}
'''

for proj in PROJECTS:
    does_md    = '\n'.join(f'- {d}' for d in proj['does'])
    how_md     = '\n'.join(f'{i+1}. {h}' for i, h in enumerate(proj['how']))
    metrics_md = '| Metric | Value |\n|--------|-------|\n'
    metrics_md += '\n'.join(f'| {k} | {v} |' for k,v in proj['metrics'])

    content = README_TEMPLATE.format(
        name=proj['name'],
        desc=proj['desc'],
        does=does_md,
        how=how_md,
        run=proj['run'],
        metrics=metrics_md,
        learned=proj['learned']
    )

    readme_path = os.path.join(proj['path'], 'README.md')
    with open(readme_path, 'w', encoding='utf-8') as f:
        f.write(content)
    print(f'Created: {readme_path}')

print('\nDone. Edit each README.md to add the demo GIF and real results.')
````

# Part 2: Creating a Demo GIF

## Why GIFs?

A GIF autoplays in GitHub README — it immediately shows the project working
without anyone having to click play or download a video.

## Method 1: Record with OBS (free screen recorder)
1. Install OBS Studio
2. Record the cv2.imshow window while running your pipeline
3. Export as MP4
4. Convert to GIF

## Method 2: Save frames from Python and make a GIF

```python
import imageio

frames = []
# ... inside your loop:
if frame_idx % 3 == 0:   # save every 3rd frame to keep GIF small
    rgb = cv2.cvtColor(display, cv2.COLOR_BGR2RGB)
    frames.append(rgb)

# After the loop:
imageio.mimsave('demo.gif', frames, fps=8)
print('GIF saved: demo.gif')
```

## Method 3: ffmpeg (best quality)
```bash
# Convert MP4 to GIF
ffmpeg -i demo.mp4 -vf 'fps=10,scale=640:-1' -loop 0 demo.gif
```

## GIF Best Practices
- Keep it under 5 MB (GitHub limit for README images)
- 8-10 FPS is smooth enough and keeps file size small
- 5-10 seconds is the right length — long enough to show the feature, short enough to be watchable
- Use a resolution of 640px wide or smaller

````python
# ── Generate demo GIF from saved pipeline frames ─────────────────────────
# Run the synthetic vehicle counter and save a GIF

import cv2, numpy as np, os

def make_road_frame(idx, h=240, w=400):
    frame = np.ones((h, w, 3), dtype=np.uint8) * 85
    frame[160:h, :] = (60,60,60)
    for x in range(0, w, 40):
        cv2.line(frame, (x,120),(x+20,120),(200,200,100),2)
    if 20 <= idx < 100:
        x = -60+(idx-20)*5
        cv2.rectangle(frame,(x,75),(x+70,115),(180,100,60),-1)
        cv2.rectangle(frame,(x+10,60),(x+60,78),(150,80,40),-1)
        cv2.circle(frame,(x+12,116),10,(30,30,30),-1)
        cv2.circle(frame,(x+58,116),10,(30,30,30),-1)
    if 50 <= idx < 130:
        x = w+20-(idx-50)*5
        cv2.rectangle(frame,(x,60),(x+100,115),(60,100,160),-1)
        cv2.circle(frame,(x+15,116),12,(30,30,30),-1)
        cv2.circle(frame,(x+85,116),12,(30,30,30),-1)
    return frame

bg_sub = cv2.createBackgroundSubtractorMOG2(history=200,varThreshold=50,detectShadows=True)
kernel = cv2.getStructuringElement(cv2.MORPH_ELLIPSE,(5,5))
for i in range(30): bg_sub.apply(make_road_frame(0))

gif_frames = []
LINE_X = 200
count = 0

for idx in range(140):
    frame = make_road_frame(idx)
    raw = bg_sub.apply(frame)
    _, fg = cv2.threshold(raw,200,255,cv2.THRESH_BINARY)
    fg = cv2.morphologyEx(fg,cv2.MORPH_OPEN,kernel)
    fg = cv2.morphologyEx(fg,cv2.MORPH_CLOSE,kernel)
    fg = cv2.dilate(fg,kernel,iterations=3)

    display = frame.copy()
    cv2.line(display,(LINE_X,0),(LINE_X,240),(0,255,255),2)
    cv2.putText(display,f'COUNT: {count}',(10,25),cv2.FONT_HERSHEY_SIMPLEX,0.7,(0,255,0),2)
    cv2.putText(display,'VEHICLE COUNTER',(10,230),cv2.FONT_HERSHEY_SIMPLEX,0.45,(200,200,200),1)

    if idx % 3 == 0:   # save every 3rd frame
        gif_frames.append(cv2.cvtColor(display, cv2.COLOR_BGR2RGB))

try:
    import imageio
    imageio.mimsave('demo.gif', gif_frames, fps=8)
    size_kb = os.path.getsize('demo.gif') / 1024
    print(f'GIF saved: demo.gif ({size_kb:.0f} KB, {len(gif_frames)} frames)')
except ImportError:
    print('imageio not installed. Run: pip install imageio')
    print(f'Would have saved {len(gif_frames)} frames as demo.gif')
````

# What I Learned

I learned how to make a project look professional on GitHub.

The most important thing is the README. A recruiter or engineer looking at your repo
will spend about 15 seconds deciding whether to look further.
If there is no README, or if the README is just a title with no explanation, they will leave.
A good README has a demo GIF at the top, a one-line description, and clear how-to-run instructions.

I learned how to create a demo GIF from Python by saving every third frame of the cv2.imshow output
using imageio. This is much simpler than recording the screen and keeps the GIF under 5 MB.

I learned about pinning package versions in requirements.txt so that anyone who clones the repo
can reproduce the exact same results. Version mismatches between numpy and opencv are a very
common cause of bugs when someone else tries to run your code.

I learned the .gitignore patterns to avoid accidentally committing large video files and model weights,
which would make the repo huge and slow to clone.

````python
# ── Phase 2 Skills Self-Assessment ──────────────────────────────────────────
# Rate yourself on each skill after completing Phase 2.
# Scale: 1 = heard of it, 3 = can use it, 5 = can teach it

import matplotlib.pyplot as plt
import numpy as np

skills = {
    # Classical ML
    'SVM + kernel trick':                3,
    'HOG / LBP features':               4,
    'PCA / Eigenfaces':                  3,
    'Sliding window + image pyramid':    3,

    # Bag of Visual Words
    'k-Means clustering':                4,
    'Bag of Visual Words':               3,
    'Image retrieval (TF-IDF)':          3,
    'Precision / Recall / mAP':          4,

    # Video Analysis
    'Background subtraction (MOG2)':     5,
    'Morphological operations':          5,
    'Contour detection':                 5,
    'Kalman filter (theory)':            4,
    'Centroid tracking':                 5,
    'Optical flow (dense)':              3,

    # Camera
    'Pinhole camera model':              4,
    'Camera calibration (checkerboard)': 3,
    'Perspective transform':             4,
    'Stereo depth estimation':           3,
}

labels = list(skills.keys())
scores = list(skills.values())
colors = ['#4e9af1' if s >= 4 else '#f4a62a' if s >= 3 else '#e05c5c' for s in scores]

fig, ax = plt.subplots(figsize=(10, 8))
bars = ax.barh(labels, scores, color=colors)
ax.set_xlim(0, 5.5)
ax.set_xlabel('Self-assessed skill level (1=heard, 3=can use, 5=can teach)')
ax.set_title('Phase 2 Skills Assessment — End of Month 2', fontsize=12, fontweight='bold')
ax.axvline(3, color='gray', ls='--', alpha=0.5, label='Can use it')
ax.axvline(5, color='green', ls='--', alpha=0.5, label='Can teach it')

# Score labels
for bar, score in zip(bars, scores):
    ax.text(score + 0.05, bar.get_y() + bar.get_height()/2,
            str(score), va='center', fontsize=9)

ax.legend()
plt.tight_layout()
plt.savefig('phase2_skills.png', dpi=100)
plt.show()

avg = sum(scores)/len(scores)
strong = [l for l,s in zip(labels,scores) if s >= 4]
weak   = [l for l,s in zip(labels,scores) if s <= 2]
print(f'Average skill level: {avg:.1f} / 5.0')
print(f'Strong areas ({len(strong)}): {strong}')
if weak:
    print(f'Areas to revisit ({len(weak)}): {weak}')
else:
    print('No areas rated below 3.')
````
